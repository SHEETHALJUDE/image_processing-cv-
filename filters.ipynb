```python
# Install OpenCV (if not already installed)
!pip install opencv-python

import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files

# ----------------------------
# Upload Image
# ----------------------------
uploaded = files.upload()

# Get uploaded file name
image_path = list(uploaded.keys())[0]

# Read image
img = cv2.imread(image_path)

# ----------------------------
# Filter Functions
# ----------------------------
def apply_mean(img):
    return cv2.blur(img, (5, 5))

def apply_gaussian(img):
    return cv2.GaussianBlur(img, (5, 5), 0)

def apply_median(img):
    return cv2.medianBlur(img, 5)

def apply_sobel(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    sobel = cv2.magnitude(sobelx, sobely)
    return cv2.convertScaleAbs(sobel)

def apply_canny(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    return cv2.Canny(gray, 100, 200)

def apply_laplacian(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    lap = cv2.Laplacian(gray, cv2.CV_64F)
    return cv2.convertScaleAbs(lap)

# ----------------------------
# Apply All Filters
# ----------------------------
def apply_all(img):
    results = {
        "Mean": apply_mean(img),
        "Gaussian": apply_gaussian(img),
        "Median": apply_median(img),
        "Sobel": apply_sobel(img),
        "Canny": apply_canny(img),
        "Laplacian": apply_laplacian(img)
    }
    return results

# ----------------------------
# Menu
# ----------------------------
while True:
    print("\n===== FILTER MENU =====")
    print("1. Mean Filter")
    print("2. Gaussian Filter")
    print("3. Median Filter")
    print("4. Sobel Edge Detection")
    print("5. Canny Edge Detection")
    print("6. Laplacian Filter")
    print("7. Apply ALL Filters")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        result = apply_mean(img)
        print("Mean Filter Output:")
        cv2_imshow(result)

    elif choice == '2':
        result = apply_gaussian(img)
        print("Gaussian Filter Output:")
        cv2_imshow(result)

    elif choice == '3':
        result = apply_median(img)
        print("Median Filter Output:")
        cv2_imshow(result)

    elif choice == '4':
        result = apply_sobel(img)
        print("Sobel Output:")
        cv2_imshow(result)

    elif choice == '5':
        result = apply_canny(img)
        print("Canny Output:")
        cv2_imshow(result)

    elif choice == '6':
        result = apply_laplacian(img)
        print("Laplacian Output:")
        cv2_imshow(result)

    elif choice == '7':
        results = apply_all(img)
        for name, res in results.items():
            print(f"{name} Filter Output:")
            cv2_imshow(res)

    elif choice == '8':
        print("Exiting...")
        break

    else:
        print("Invalid choice!")
```

```python
# Install OpenCV (if needed)
!pip install opencv-python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ----------------------------
# Upload Image
# ----------------------------
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
img = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ----------------------------
# Individual Filters
# ----------------------------
mean = cv2.blur(img, (5, 5))
gaussian = cv2.GaussianBlur(img, (5, 5), 0)
median = cv2.medianBlur(img, 5)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel = cv2.magnitude(sobelx, sobely)
sobel = cv2.convertScaleAbs(sobel)

laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

# Convert BGR → RGB for display
mean_rgb = cv2.cvtColor(mean, cv2.COLOR_BGR2RGB)
gaussian_rgb = cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)
median_rgb = cv2.cvtColor(median, cv2.COLOR_BGR2RGB)

# ----------------------------
# Apply ALL filters sequentially
# ----------------------------
combined = cv2.blur(img, (5, 5))
combined = cv2.GaussianBlur(combined, (5, 5), 0)
combined = cv2.medianBlur(combined, 5)

gray_comb = cv2.cvtColor(combined, cv2.COLOR_BGR2GRAY)

sobelx_c = cv2.Sobel(gray_comb, cv2.CV_64F, 1, 0, ksize=3)
sobely_c = cv2.Sobel(gray_comb, cv2.CV_64F, 0, 1, ksize=3)
combined = cv2.magnitude(sobelx_c, sobely_c)
combined = cv2.convertScaleAbs(combined)

# ----------------------------
# Display All Images
# ----------------------------
plt.figure(figsize=(14, 10))

plt.subplot(3, 3, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis('off')

plt.subplot(3, 3, 2)
plt.imshow(mean_rgb)
plt.title("Mean Filter")
plt.axis('off')

plt.subplot(3, 3, 3)
plt.imshow(gaussian_rgb)
plt.title("Gaussian Filter")
plt.axis('off')

plt.subplot(3, 3, 4)
plt.imshow(median_rgb)
plt.title("Median Filter")
plt.axis('off')

plt.subplot(3, 3, 5)
plt.imshow(sobel, cmap='gray')
plt.title("Sobel Filter")
plt.axis('off')

plt.subplot(3, 3, 6)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Filter")
plt.axis('off')

plt.subplot(3, 3, 8)
plt.imshow(combined, cmap='gray')
plt.title("All Filters Combined")
plt.axis('off')

plt.tight_layout()
plt.show()
```
